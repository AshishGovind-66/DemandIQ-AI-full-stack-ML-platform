## 1. Run Inventory Optimization Policy


```python
df = pd.read_csv(settings.DATA_SAMPLE_DIR / "retail_sales_sample.csv")
req = InventoryOptimizationRequest(
    dataset_id="retail-sample",
    lead_time_days=7,
    service_level=0.95,
    default_current_stock=15,
    min_order_quantity=10,
    pack_size=5
)
summary = InventoryOptimizationEngine.optimize(df, req)
print(f"Total Products: {summary.total_products_analyzed}")
print(f"Stockout Risks: {summary.items_at_stockout_risk}")
print(f"Suggested Order Units: {summary.total_suggested_reorder_units}")

recs_df = pd.DataFrame([r.dict() for r in summary.recommendations])
print(recs_df[["product_id", "daily_demand_mean", "safety_stock", "reorder_point", "suggested_order_qty", "stockout_risk"]].head(10))
```
