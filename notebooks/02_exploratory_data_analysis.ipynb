```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, str(Path(".." if Path(".").resolve().name == "notebooks" else ".").resolve()))
from backend.app.services.eda_service import EDAService
from backend.app.core.config import settings
```

## 1. Load Data & Compute Profiling


```python
data_path = settings.DATA_SAMPLE_DIR / "retail_sales_sample.csv"
df = pd.read_csv(data_path)
df["date"] = pd.to_datetime(df["date"])
print("Loaded rows:", len(df))
print(df.describe())
```

## 2. Daily Sales Trend & Seasonality Visualization


```python
plt.figure(figsize=(12, 5))
daily_total = df.groupby("date")["sales"].sum()
plt.plot(daily_total.index, daily_total.values, color="#2563eb", lw=1.5)
plt.title("Total Daily Retail Demand Over Time", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Day of Week Pattern (Weekly Seasonality)


```python
df["day_name"] = df["date"].dt.day_name()
dow_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x="day_name", y="sales", order=dow_order, palette="Blues_d")
plt.title("Mean Sales Volume by Day of Week", fontsize=13, fontweight="bold")
plt.xlabel("Day of Week")
plt.ylabel("Mean Sales")
plt.tight_layout()
plt.show()
```
