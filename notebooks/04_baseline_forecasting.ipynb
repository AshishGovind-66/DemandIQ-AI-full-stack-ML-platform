# DemandIQ AI - Notebook 04: Baseline Forecasting Models
Evaluates Naive (lag_1) and Seasonal Naive (lag_7) baselines as benchmark comparison standards.


```python
import sys
from pathlib import Path
import pandas as pd

sys.path.insert(0, str(Path(".." if Path(".").resolve().name == "notebooks" else ".").resolve()))
from backend.app.forecasting.baseline_models import NaiveForecaster, SeasonalNaiveForecaster
from backend.app.utils.metrics import compute_all_metrics
from backend.app.preprocessing.feature_engineering import FeatureEngineer
from backend.app.utils.time_series import chronological_split
from backend.app.core.config import settings
```

## 1. Prepare Validation & Test Data


```python
df = pd.read_csv(settings.DATA_SAMPLE_DIR / "retail_sales_sample.csv")
train, val, test = chronological_split(df, validation_days=14, test_days=14)
fe = FeatureEngineer()
fe.fit_transform(train)
test_feat = fe.transform(test)
X_test, y_test = fe.get_features_and_target(test_feat)
```

## 2. Benchmark Naive & Seasonal Naive Baselines


```python
naive = NaiveForecaster().fit(X_test, y_test)
naive_preds = naive.predict(X_test)
print("Naive Baseline Metrics:", compute_all_metrics(y_test, naive_preds))

snaive = SeasonalNaiveForecaster(season_length=7).fit(X_test, y_test)
snaive_preds = snaive.predict(X_test)
print("Seasonal Naive (7d) Metrics:", compute_all_metrics(y_test, snaive_preds))
```
